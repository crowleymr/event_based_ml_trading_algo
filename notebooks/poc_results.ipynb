# Slice 1 generated report explorer
This notebook reads versioned report outputs only. It never trains, selects, or changes the immutable source run. Set TRADING_REPORT_DIR to a generated v1 report directory before running in another environment.

```python
from pathlib import Path
import json
import os
import polars as pl

report_dir = Path(os.environ.get(
    'TRADING_REPORT_DIR',
    '../reports/20260912T071137Z-9899fd9a/v1',
))
if not (report_dir / 'provenance.json').is_file():
    raise FileNotFoundError(
        'Generate the report first or set TRADING_REPORT_DIR to its v1 directory.'
    )
provenance = json.loads((report_dir / 'provenance.json').read_text())
comparison = pl.read_parquet(report_dir / 'experiment_comparison.parquet')
provenance['source_run_id'], comparison
```

```python
test_results = comparison.filter(pl.col('split') == 'test').select(
    'experiment_id', 'display_label', 'total_return', 'sharpe',
    'maximum_drawdown', 'mean_ic', 'rmse'
)
test_results
```
